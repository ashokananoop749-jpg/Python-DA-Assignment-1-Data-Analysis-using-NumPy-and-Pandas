```python
import pandas as pd
```

```python
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
```

```python
rank1_mark = marks.iloc[0] # or marks[0]
top3_marks = marks.loc['Rank1':'Rank3']
rank3_mark = marks.iloc[2]
above_90 = marks[marks > 90]

print("1st Rank Mark:", rank1_mark)
```

```python
print("Top 3 Ranks:\n", top3_marks)
```

```python
print("3rd Rank Mark:", rank3_mark)
```

```python
print("Marks > 90:\n", above_90)
```

```python
marks['Rank1'] = 100
marks = marks.drop('Rank5')
cgpa = marks / 10

print("\nUpdated Marks:\n", marks)
```

```python
print("CGPA:\n", cgpa)
```

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                       'Clothing', 'Electronics', 'Furniture', 'Clothing',
                       'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
print("--- DataFrame Overview ---")
```

```python
print(transactions)
```

```python
print("\nHead:\n", transactions.head())
```

```python
print("\nTail:\n", transactions.tail())
```

```python
print("\nShape:", transactions.shape)
```

```python
print("Column Names:", transactions.columns.tolist())
```

```python
print("Data Types:\n", transactions.dtypes)
```

```python
print("\nCategory & Amount:\n", transactions[['ProductCategory', 'Amount']])
```

```python
print("\nLast 3 Columns:\n", transactions.iloc[:, -3:])
```

```python
north_high = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("\nNorth Region with Amount > 200:\n", north_high)
```

```python
print("\nValue Counts (ProductCategory):\n", transactions['ProductCategory'].value_counts())
```

```python
print("\nUnique Regions:", transactions['Region'].unique())
```

```python
print("\nMean Amount by Region:\n", transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10
```

```python
transactions = transactions[transactions['TransactionID'] != 109]
```

```python
transactions = transactions.drop(columns=['Discount'])
```

```python
print("\n--- Final DataFrame ---")
```
